```python
s1 = {1, 2, 3, 4, 5}
s2 = {4, 5, 6, 7, 8}

s3 = s1 & s2
s4 = s1 | s2
s5 = s1 - s2
s6 = s1 ^ s2

print("s3:", s3)
print("s4:", s4)
print("s5:", s5)
print("s6:", s6)
```
